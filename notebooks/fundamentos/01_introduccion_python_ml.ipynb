## 2. Comprensión de Listas (List Comprehension)

La comprensión de listas es una forma elegante y eficiente de crear listas en Python. Es muy útil en ML para transformar datos.

### Sintaxis:
```python
nueva_lista = [expresion for elemento in iterable if condicion]
```

````python
# Ejemplo: Crear una lista de cuadrados
numeros = [1, 2, 3, 4, 5]

# Forma tradicional
cuadrados_tradicional = []
for n in numeros:
    cuadrados_tradicional.append(n ** 2)

# Con list comprehension (más elegante)
cuadrados = [n ** 2 for n in numeros]

print(f"Cuadrados: {cuadrados}")
````

````python
# Ejemplo con condición: Solo números pares
numeros = range(1, 11)  # 1 a 10
pares = [n for n in numeros if n % 2 == 0]
print(f"Números pares: {pares}")
````

## 5. Clases y Programación Orientada a Objetos

Las clases nos permiten crear nuestros propios tipos de datos. En ML, todos los algoritmos se implementan como clases.

### Conceptos Clave:
- Una clase es un "plano" para crear objetos
- Los objetos tienen atributos (datos) y métodos (funciones)
- `__init__` es el constructor (se ejecuta al crear un objeto)
- `self` se refiere a la instancia actual del objeto

````python
# Ejemplo: Clase para un conjunto de datos simple
class Dataset:
    def __init__(self, nombre, datos):
        """Constructor de la clase"""
        self.nombre = nombre
        self.datos = datos
    
    def obtener_tamaño(self):
        """Retorna el número de elementos en el dataset"""
        return len(self.datos)
    
    def obtener_media(self):
        """Retorna la media de los datos"""
        return sum(self.datos) / len(self.datos)
    
    def __str__(self):
        """Representación en string del objeto"""
        return f"Dataset(nombre='{self.nombre}', tamaño={self.obtener_tamaño()})"

# Crear una instancia de la clase
mi_dataset = Dataset('Temperaturas', [20, 22, 19, 25, 23])

# Usar los métodos
print(mi_dataset)
print(f"Tamaño: {mi_dataset.obtener_tamaño()}")
print(f"Media: {mi_dataset.obtener_media()}")
````

## 6. Lambda Functions y Map/Filter

Las funciones lambda son funciones anónimas de una sola línea. Son muy útiles para operaciones rápidas.

### Funciones de orden superior:
- `map()`: Aplica una función a cada elemento de un iterable
- `filter()`: Filtra elementos basándose en una condición

````python
# Ejemplo de lambda
cuadrado = lambda x: x ** 2
print(f"Cuadrado de 5: {cuadrado(5)}")

# Usando map()
numeros = [1, 2, 3, 4, 5]
cuadrados = list(map(lambda x: x ** 2, numeros))
print(f"Cuadrados con map: {cuadrados}")

# Usando filter()
pares = list(filter(lambda x: x % 2 == 0, numeros))
print(f"Números pares con filter: {pares}")
````
