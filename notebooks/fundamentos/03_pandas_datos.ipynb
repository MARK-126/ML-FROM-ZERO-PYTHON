## 1. Series de Pandas

Una **Series** es un array unidimensional etiquetado.

```python
# Crear una Series
temperaturas = pd.Series([20, 22, 19, 25, 23], 
                        index=['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes'])

print(temperaturas)
print(f"\nTipo: {type(temperaturas)}")
print(f"\nTemperatura del Miércoles: {temperaturas['Miércoles']}")
```

## 2. DataFrame - La Estructura Principal

## 3. Selección de Datos

Pandas ofrece múltiples formas de seleccionar datos.

```python
# Seleccionar una columna
print("Columna 'Nombre':")
print(df['Nombre'])

# Seleccionar múltiples columnas
print("\nColumnas 'Nombre' y 'Salario':")
print(df[['Nombre', 'Salario']])
```

```python
# Seleccionar por posición con iloc (integer location)
print("Primera fila:")
print(df.iloc[0])

print("\nPrimeras 2 filas y primeras 2 columnas:")
print(df.iloc[:2, :2])
```

## 4. Filtrado de Datos

Filtrar datos basándose en condiciones es fundamental en ML.

```python
# Filtrar filas que cumplan una condición
print("Personas con edad > 25:")
mayores_25 = df[df['Edad'] > 25]
print(mayores_25)

print("\nPersonas de Madrid:")
madrid = df[df['Ciudad'] == 'Madrid']
print(madrid)
```

```python
# Múltiples condiciones
print("Personas de Madrid con edad > 25:")
filtro = df[(df['Ciudad'] == 'Madrid') & (df['Edad'] > 25)]
print(filtro)

print("\nPersonas de Madrid O Barcelona:")
ciudades = df[(df['Ciudad'] == 'Madrid') | (df['Ciudad'] == 'Barcelona')]
print(ciudades)
```

## 5. Datos Faltantes

Manejar datos faltantes (NaN) es crucial en ML.

```python
# Crear DataFrame con valores faltantes
datos_faltantes = {
    'A': [1, 2, np.nan, 4],
    'B': [5, np.nan, np.nan, 8],
    'C': [9, 10, 11, 12]
}
df_nan = pd.DataFrame(datos_faltantes)
print("DataFrame con valores faltantes:")
print(df_nan)

# Detectar valores faltantes
print("\n¿Hay valores NaN?")
print(df_nan.isna())

print("\nCantidad de NaN por columna:")
print(df_nan.isna().sum())
```

```python
# Eliminar filas con NaN
print("Eliminar filas con NaN:")
print(df_nan.dropna())

# Rellenar NaN con un valor
print("\nRellenar NaN con 0:")
print(df_nan.fillna(0))

# Rellenar con la media de la columna
print("\nRellenar NaN con la media:")
print(df_nan.fillna(df_nan.mean()))
```

## 6. Operaciones y Transformaciones

```python
# Agregar una nueva columna
df['Salario_Anual_Miles'] = df['Salario'] / 1000
print(df)

# Aplicar una función a una columna
df['Edad_Grupo'] = df['Edad'].apply(lambda x: 'Joven' if x < 30 else 'Mayor')
print("\nCon grupos de edad:")
print(df)
```

```python
# Ordenar datos
print("Ordenar por salario (descendente):")
print(df.sort_values('Salario', ascending=False))

print("\nOrdenar por múltiples columnas:")
print(df.sort_values(['Ciudad', 'Edad']))
```

## 7. Agrupación (GroupBy)

GroupBy es una herramienta poderosa para análisis agregados.

```python
# Agrupar por ciudad y calcular estadísticas
print("Salario promedio por ciudad:")
print(df.groupby('Ciudad')['Salario'].mean())

print("\nEstadísticas completas por ciudad:")
print(df.groupby('Ciudad')['Salario'].describe())
```

```python
# Múltiples agregaciones
print("Múltiples estadísticas:")
print(df.groupby('Ciudad').agg({
    'Edad': ['mean', 'min', 'max'],
    'Salario': ['mean', 'sum']
}))
```

## 8. Combinar DataFrames

```python
# Merge (similar a SQL JOIN)
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Nombre': ['Ana', 'Luis', 'María']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 4],
    'Salario': [30000, 45000, 35000]
})

print("df1:")
print(df1)
print("\ndf2:")
print(df2)

# Inner join
merged = pd.merge(df1, df2, on='ID', how='inner')
print("\nInner Join:")
print(merged)

# Left join
left_merged = pd.merge(df1, df2, on='ID', how='left')
print("\nLeft Join:")
print(left_merged)
```

```python
# Concatenar DataFrames
df_a = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df_b = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

# Concatenar verticalmente (apilar filas)
concatenado = pd.concat([df_a, df_b], ignore_index=True)
print("Concatenación vertical:")
print(concatenado)
```
