## 2 - Listas y Operaciones Básicas

Las listas son una de las estructuras de datos más fundamentales en Python. En Machine Learning, las usaremos constantemente para:

* Almacenar datasets y características
* Guardar predicciones de modelos
* Mantener historial de métricas durante el entrenamiento
* Organizar lotes de datos (batches)

### Propiedades de las Listas:

* **Ordenadas**: Los elementos mantienen el orden de inserción
* **Mutables**: Pueden modificarse después de su creación
* **Heterogéneas**: Pueden contener elementos de diferentes tipos
* **Indexables**: Acceso directo por índice (empezando en 0)

**Notación**: Para una lista $L = [l_1, l_2, ..., l_n]$:
* Longitud: $|L| = n$
* Suma: $\sum_{i=1}^{n} l_i$
* Promedio: $\bar{L} = \frac{1}{n}\sum_{i=1}^{n} l_i$

````python
# Demostración: Operaciones básicas con listas
numeros = [1, 2, 3, 4, 5]

print("Lista original:", numeros)
print("Primer elemento (índice 0):", numeros[0])
print("Último elemento (índice -1):", numeros[-1])
print("Sublista (slicing) [1:4]:", numeros[1:4])
print("Longitud:", len(numeros))
print("Suma de elementos:", sum(numeros))
print("Promedio:", sum(numeros) / len(numeros))
````

````python
# Métodos comunes de listas
numeros = [1, 2, 3]

# Agregar elementos
numeros.append(4)          # Agregar al final
print("Después de append(4):", numeros)

numeros.insert(0, 0)       # Insertar en posición específica
print("Después de insert(0, 0):", numeros)

numeros.extend([5, 6])     # Extender con otra lista
print("Después de extend([5, 6]):", numeros)

# Eliminar elementos
numeros.remove(0)          # Eliminar por valor
print("Después de remove(0):", numeros)

ultimo = numeros.pop()     # Eliminar y retornar último elemento
print(f"Después de pop() (removido {ultimo}):", numeros)
````

## 3 - List Comprehension

List comprehension es una forma elegante y eficiente de crear listas en Python. Es especialmente útil en Machine Learning para transformar datos.

### Sintaxis General:

```python
nueva_lista = [expresion for elemento in iterable if condicion]
```

### Ventajas:

* **Más rápida** que bucles for tradicionales
* **Más legible** y concisa
* **Más Pythonic** (idiomática)

### Comparación de Rendimiento:

Para una lista de $n$ elementos:
* Bucle tradicional: $O(n)$ tiempo + overhead de append
* List comprehension: $O(n)$ tiempo optimizado

**Notación Matemática:** Si queremos aplicar una función $f$ a cada elemento de un conjunto:
$$S_{nuevo} = \{f(x) \mid x \in S_{original}\}$$

````python
# Ejemplo: Comparación entre bucle tradicional y list comprehension
numeros = [1, 2, 3, 4, 5]

# Método 1: Bucle tradicional (verbose)
cuadrados_tradicional = []
for n in numeros:
    cuadrados_tradicional.append(n ** 2)

# Método 2: List comprehension (elegante)
cuadrados_comprehension = [n ** 2 for n in numeros]

print("Método tradicional:", cuadrados_tradicional)
print("List comprehension:", cuadrados_comprehension)
print("¿Son iguales?", cuadrados_tradicional == cuadrados_comprehension)
````

````python
# Ejemplo: List comprehension con condición
numeros = range(1, 11)  # 1 a 10

# Solo números pares
pares = [n for n in numeros if n % 2 == 0]
print("Números pares:", pares)

# Solo impares al cuadrado
impares_cuadrado = [n ** 2 for n in numeros if n % 2 != 0]
print("Impares al cuadrado:", impares_cuadrado)

# Transformación condicional (if-else)
clasificacion = ["par" if n % 2 == 0 else "impar" for n in numeros]
print("Clasificación:", clasificacion)
````

## 4 - Diccionarios

Los diccionarios almacenan pares clave-valor y son fundamentales en Machine Learning para:

* Representar parámetros de modelos (pesos, bias, etc.)
* Almacenar configuraciones e hiperparámetros
* Guardar métricas de evaluación
* Mantener cachés de resultados intermedios

### Propiedades de los Diccionarios:

* **No ordenados** (hasta Python 3.6) / **Ordenados** (Python 3.7+)
* **Mutables**: Pueden modificarse después de su creación
* **Claves únicas**: Cada clave puede aparecer solo una vez
* **Acceso rápido**: Búsqueda en tiempo $O(1)$ promedio

### Complejidad Temporal:

| Operación | Complejidad |
|-----------|-------------|
| Acceso por clave | $O(1)$ |
| Inserción | $O(1)$ |
| Eliminación | $O(1)$ |
| Búsqueda de valor | $O(n)$ |

````python
# Demostración: Diccionario de parámetros de un modelo
modelo_params = {
    'learning_rate': 0.01,
    'epochs': 100,
    'batch_size': 32,
    'optimizer': 'adam',
    'layers': [64, 32, 10]
}

print("Parámetros del modelo:")
for clave, valor in modelo_params.items():
    print(f"  {clave}: {valor}")

print(f"\nLearning rate: {modelo_params['learning_rate']}")
print(f"Número de capas: {len(modelo_params['layers'])}")
````

````python
# Ejemplo: Función con docstring profesional
def calcular_media(numeros):
    """
    Calcula la media aritmética de una lista de números.
    
    La media se define como:
    $$\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i$$
    
    Parameters:
    -----------
    numeros : list or array-like
        Lista de números (int o float)
    
    Returns:
    --------
    float
        Media aritmética de los números
    
    Raises:
    -------
    ValueError
        Si la lista está vacía
    
    Examples:
    ---------
    >>> calcular_media([1, 2, 3, 4, 5])
    3.0
    
    >>> calcular_media([10, 20])
    15.0
    """
    if len(numeros) == 0:
        raise ValueError("La lista no puede estar vacía")
    
    return sum(numeros) / len(numeros)


# Probar la función
datos = [10, 20, 30, 40, 50]
media = calcular_media(datos)
print(f"Media de {datos}: {media}")
````

## 6 - Clases y Programación Orientada a Objetos

Las clases nos permiten crear nuestros propios tipos de datos. En Machine Learning, **todos los algoritmos se implementan como clases** (scikit-learn, TensorFlow, PyTorch).

### Conceptos Clave de POO:

* **Clase**: Plantilla para crear objetos
* **Objeto**: Instancia de una clase
* **Atributos**: Variables que pertenecen al objeto (estado)
* **Métodos**: Funciones que pertenecen al objeto (comportamiento)
* **Constructor (`__init__`)**: Inicializa el objeto
* **`self`**: Referencia a la instancia actual

### Estructura de una Clase:

```python
class MiClase:
    def __init__(self, param1, param2):
        self.atributo1 = param1
        self.atributo2 = param2
    
    def metodo(self):
        return self.atributo1 + self.atributo2
```

````python
# Ejemplo: Clase para representar un dataset
class Dataset:
    """
    Clase para representar y analizar un conjunto de datos.
    
    Attributes:
    -----------
    nombre : str
        Nombre del dataset
    datos : list
        Lista de valores numéricos
    """
    
    def __init__(self, nombre, datos):
        """Inicializa el dataset con nombre y datos."""
        self.nombre = nombre
        self.datos = datos
    
    def obtener_tamaño(self):
        """Retorna el número de elementos en el dataset."""
        return len(self.datos)
    
    def obtener_media(self):
        """Retorna la media de los datos."""
        return sum(self.datos) / len(self.datos)
    
    def obtener_maximo(self):
        """Retorna el valor máximo."""
        return max(self.datos)
    
    def obtener_minimo(self):
        """Retorna el valor mínimo."""
        return min(self.datos)
    
    def __str__(self):
        """Representación en string del objeto."""
        return f"Dataset(nombre='{self.nombre}', tamaño={self.obtener_tamaño()})"
    
    def __repr__(self):
        """Representación técnica del objeto."""
        return self.__str__()


# Crear una instancia
temperaturas = Dataset('Temperaturas Enero', [20, 22, 19, 25, 23, 21])

# Usar los métodos
print(temperaturas)
print(f"Tamaño: {temperaturas.obtener_tamaño()}")
print(f"Media: {temperaturas.obtener_media():.1f}°C")
print(f"Máxima: {temperaturas.obtener_maximo()}°C")
print(f"Mínima: {temperaturas.obtener_minimo()}°C")
````

## 7 - Lambda Functions y Map/Filter

Las funciones lambda son funciones anónimas de una sola línea, muy útiles en Machine Learning para:

* Transformaciones rápidas de datos
* Funciones de activación personalizadas
* Callbacks y funciones de ordenamiento

### Sintaxis de Lambda:

```python
lambda argumentos: expresion
```

### Funciones de Orden Superior:

* **`map(funcion, iterable)`**: Aplica una función a cada elemento
* **`filter(funcion, iterable)`**: Filtra elementos que cumplen una condición
* **`reduce(funcion, iterable)`**: Reduce la secuencia a un solo valor

**Notación Funcional:**
* Map: $map(f, [x_1, x_2, ..., x_n]) \rightarrow [f(x_1), f(x_2), ..., f(x_n)]$
* Filter: $filter(p, S) \rightarrow \{x \in S \mid p(x) = True\}$

````python
# Ejemplo: Funciones lambda básicas
cuadrado = lambda x: x ** 2
suma = lambda x, y: x + y
es_par = lambda x: x % 2 == 0

print(f"Cuadrado de 5: {cuadrado(5)}")
print(f"Suma de 3 y 4: {suma(3, 4)}")
print(f"¿Es 6 par? {es_par(6)}")
print(f"¿Es 7 par? {es_par(7)}")
````

````python
# Ejemplo: Usando map()
numeros = [1, 2, 3, 4, 5]

# Cuadrados con map
cuadrados = list(map(lambda x: x ** 2, numeros))
print(f"Cuadrados con map: {cuadrados}")

# Aplicar función a múltiples listas
lista1 = [1, 2, 3]
lista2 = [10, 20, 30]
sumas = list(map(lambda x, y: x + y, lista1, lista2))
print(f"Sumas elemento a elemento: {sumas}")
````

````python
# Ejemplo: Usando filter()
numeros = range(1, 11)  # 1 a 10

# Filtrar pares
pares = list(filter(lambda x: x % 2 == 0, numeros))
print(f"Números pares: {pares}")

# Filtrar números mayores a 5
mayores_5 = list(filter(lambda x: x > 5, numeros))
print(f"Mayores a 5: {mayores_5}")

# Combinar filter y map
cuadrados_impares = list(map(lambda x: x ** 2, 
                              filter(lambda x: x % 2 != 0, numeros)))
print(f"Cuadrados de impares: {cuadrados_impares}")
````
