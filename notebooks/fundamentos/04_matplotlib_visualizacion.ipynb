# 📈 Notebook 04: Matplotlib - Visualización de Datos

## Objetivos
- Crear gráficos básicos con Matplotlib
- Personalizar visualizaciones
- Crear gráficos para análisis de ML
- Usar Seaborn para visualizaciones avanzadas

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
%matplotlib inline
```

## 1. Gráfico de Líneas

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, label='sin(x)', color='blue', linewidth=2)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Función Seno')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 2. Scatter Plot

```python
np.random.seed(42)
x = np.random.randn(100)
y = 2 * x + np.random.randn(100) * 0.5

plt.figure(figsize=(10, 6))
plt.scatter(x, y, alpha=0.6, edgecolors='k')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot - Correlación')
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Histogramas

```python
datos = np.random.randn(1000)

plt.figure(figsize=(10, 6))
plt.hist(datos, bins=30, edgecolor='black', alpha=0.7)
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title('Histograma - Distribución Normal')
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Subplots - Múltiples Gráficos

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

x = np.linspace(0, 10, 100)

# Subplot 1
axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title('sin(x)')

# Subplot 2
axes[0, 1].plot(x, np.cos(x), color='red')
axes[0, 1].set_title('cos(x)')

# Subplot 3
axes[1, 0].plot(x, x**2, color='green')
axes[1, 0].set_title('x²')

# Subplot 4
axes[1, 1].plot(x, np.log(x + 1), color='orange')
axes[1, 1].set_title('log(x+1)')

plt.tight_layout()
plt.show()
```

## 5. Seaborn - Visualizaciones Estadísticas

```python
# Crear datos de ejemplo
df = pd.DataFrame({
    'X': np.random.randn(100),
    'Y': np.random.randn(100),
    'Categoria': np.random.choice(['A', 'B', 'C'], 100)
})

# Scatter plot con categorías
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='X', y='Y', hue='Categoria', s=100)
plt.title('Scatter Plot con Categorías')
plt.show()
```
