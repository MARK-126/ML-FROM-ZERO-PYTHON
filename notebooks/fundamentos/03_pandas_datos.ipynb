# Crear una Series con índices personalizados

                        index=['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes'])

print("📊 Series de temperaturas:")
print(temperaturas)
print(f"\n🔍 Tipo: {type(temperaturas)}")
print(f"🌡️ Temperatura del Miércoles: {temperaturas['Miércoles']}°C")
print(f"📏 Longitud: {len(temperaturas)}")

````python
### 2.2 - DataFrame

Un **DataFrame** es una tabla bidimensional con filas y columnas etiquetadas, similar a una hoja de cálculo o tabla SQL.

**Estructura:**
- Cada columna es una Series
- Todas las columnas comparten el mismo índice
- Puede contener diferentes tipos de datos por columna
````

El filtrado de datos usando condiciones booleanas es fundamental en Machine Learning para:

````python
# Filtrar filas que cumplan una condición
print("Personas con edad > 25:")
mayores_25 = df[df['Edad'] > 25]
print(mayores_25)

print("\nPersonas de Madrid:")
madrid = df[df['Ciudad'] == 'Madrid']
print(madrid)
````

````python
# Múltiples condiciones
print("Personas de Madrid con edad > 25:")
filtro = df[(df['Ciudad'] == 'Madrid') & (df['Edad'] > 25)]
print(filtro)

print("\nPersonas de Madrid O Barcelona:")
ciudades = df[(df['Ciudad'] == 'Madrid') | (df['Ciudad'] == 'Barcelona')]
print(ciudades)
````

## 7 - Agrupación (GroupBy)

**GroupBy** es una de las operaciones más poderosas en Pandas, basada en el paradigma **"split-apply-combine"**:

1. **Split**: Dividir los datos en grupos basados en criterios
2. **Apply**: Aplicar una función a cada grupo independientemente  
3. **Combine**: Combinar los resultados en una estructura de datos

**Operación matemática:**

Para una columna $X$ agrupada por categoría $g$:

$$\text{mean}_g(X) = \frac{1}{|G_g|} \sum_{i \in G_g} x_i$$

donde $G_g$ es el conjunto de índices del grupo $g$.

````python
# Múltiples agregaciones
print("Múltiples estadísticas:")
print(df.groupby('Ciudad').agg({
    'Edad': ['mean', 'min', 'max'],
    'Salario': ['mean', 'sum']
}))
````

## 8 - Combinar DataFrames

Combinar múltiples DataFrames es esencial en Machine Learning para:
- Integrar datos de múltiples fuentes
- Agregar features adicionales
- Unir datos de entrenamiento con metadatos

### 8.1 - Merge (similar a SQL JOIN)

**Tipos de merge:**
- **Inner**: Solo filas con claves en ambos DataFrames
- **Left**: Todas las filas del DataFrame izquierdo
- **Right**: Todas las filas del DataFrame derecho
- **Outer**: Todas las filas de ambos DataFrames

````python
# Concatenar DataFrames
df_a = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df_b = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

# Concatenar verticalmente (apilar filas)
concatenado = pd.concat([df_a, df_b], ignore_index=True)
print("Concatenación vertical:")
print(concatenado)
````
