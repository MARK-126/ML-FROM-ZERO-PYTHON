## 3 - Scatter Plots (Gráficos de Dispersión)

Los **scatter plots** son fundamentales en ML para:
- Visualizar relaciones entre variables
- Identificar correlaciones
- Detectar outliers y clusters
- Evaluar predicciones vs valores reales

**Aplicación en ML:**
$$y = f(X) + \epsilon$$

donde visualizamos la relación entre features $X$ y target $y$, más el ruido $\epsilon$.

# Datos con correlación lineal + ruido
np.random.seed(42)
n_samples = 100
x = np.random.randn(n_samples)
y = 2 * x + np.random.randn(n_samples) * 0.5  # y = 2x + ruido

plt.figure(figsize=(10, 6))
plt.scatter(x, y, alpha=0.6, edgecolors='k', s=50)
plt.xlabel('Feature X', fontsize=12)
plt.ylabel('Target Y', fontsize=12)
plt.title('Scatter Plot - Relación Lineal con Ruido', fontsize=14, fontweight='bold')

# Agregar línea de tendencia
z = np.polyfit(x, y, 1)
p = np.poly1d(z)
plt.plot(x, p(x), "r--", linewidth=2, label=f'Tendencia: y={z[0]:.2f}x+{z[1]:.2f}')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

## 4 - Histogramas

Los **histogramas** visualizan la distribución de una variable:
- Frecuencia de valores en rangos (bins)
- Identificar distribución: normal, uniforme, sesgada
- Detectar outliers y multimodalidad

**Distribución Normal:**
$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2}$$

donde $\mu$ es la media y $\sigma$ la desviación estándar.

# Generar datos de distribución normal
np.random.seed(42)
datos = np.random.randn(1000)

plt.figure(figsize=(10, 6))
n, bins, patches = plt.hist(datos, bins=30, edgecolor='black', alpha=0.7, color='skyblue')

plt.xlabel('Valor', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)
plt.title('Histograma - Distribución Normal (μ=0, σ=1)', fontsize=14, fontweight='bold')

# Agregar líneas para media y desviación estándar
plt.axvline(datos.mean(), color='red', linestyle='--', linewidth=2, label=f'Media: {datos.mean():.2f}')
plt.axvline(datos.mean() + datos.std(), color='green', linestyle='--', linewidth=2, label=f'μ+σ: {datos.mean() + datos.std():.2f}')
plt.axvline(datos.mean() - datos.std(), color='green', linestyle='--', linewidth=2, label=f'μ-σ: {datos.mean() - datos.std():.2f}')

plt.legend(fontsize=11)
plt.grid(True, alpha=0.3, axis='y')
plt.show()

## 5 - Subplots (Múltiples Gráficos)

**Subplots** permiten crear múltiples gráficos en una sola figura para:
- Comparar diferentes visualizaciones
- Mostrar múltiples aspectos de los datos
- Crear dashboards informativos

**Sintaxis:**
```python
fig, axes = plt.subplots(nrows, ncols, figsize=(ancho, alto))
```

donde `axes` es un array de objetos Axes para cada subplot.

# Crear una cuadrícula 2x2 de gráficos
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

x = np.linspace(0, 10, 100)

# Subplot 1: Función seno
axes[0, 0].plot(x, np.sin(x), color='blue', linewidth=2)
axes[0, 0].set_title('sin(x)', fontsize=12, fontweight='bold')
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].set_xlabel('x')
axes[0, 0].set_ylabel('y')

# Subplot 2: Función coseno
axes[0, 1].plot(x, np.cos(x), color='red', linewidth=2)
axes[0, 1].set_title('cos(x)', fontsize=12, fontweight='bold')
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].set_xlabel('x')
axes[0, 1].set_ylabel('y')

# Subplot 3: Función exponencial
axes[1, 0].plot(x, np.exp(-x/5), color='green', linewidth=2)
axes[1, 0].set_title('exp(-x/5)', fontsize=12, fontweight='bold')
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].set_xlabel('x')
axes[1, 0].set_ylabel('y')

# Subplot 4: Función logarítmica
axes[1, 1].plot(x + 1, np.log(x + 1), color='orange', linewidth=2)
axes[1, 1].set_title('log(x+1)', fontsize=12, fontweight='bold')
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].set_xlabel('x')
axes[1, 1].set_ylabel('y')

plt.tight_layout()  # Ajustar automáticamente el espaciado
plt.show()

# Crear datos de ejemplo con categorías
np.random.seed(42)
n_per_category = 100

df = pd.DataFrame({
    'Feature_1': np.concatenate([
        np.random.randn(n_per_category) + 0,
        np.random.randn(n_per_category) + 3,
        np.random.randn(n_per_category) + 6
    ]),
    'Feature_2': np.concatenate([
        np.random.randn(n_per_category) + 0,
        np.random.randn(n_per_category) + 2,
        np.random.randn(n_per_category) + 4
    ]),
    'Clase': ['A'] * n_per_category + ['B'] * n_per_category + ['C'] * n_per_category
})

# Scatter plot con seaborn
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Feature_1', y='Feature_2', hue='Clase', s=100, alpha=0.7, edgecolor='k')
plt.title('Scatter Plot con Clases - Problema de Clasificación', fontsize=14, fontweight='bold')
plt.xlabel('Feature 1', fontsize=12)
plt.ylabel('Feature 2', fontsize=12)
plt.legend(title='Clase', fontsize=11)
plt.grid(True, alpha=0.3)

### 7.1 - Curva de Aprendizaje (Learning Curve)

Las **curvas de aprendizaje** visualizan el rendimiento del modelo vs cantidad de datos de entrenamiento:
- Diagnosticar underfitting o overfitting
- Decidir si necesitamos más datos
- Evaluar convergencia del entrenamiento

plt.figure(figsize=(10, 6))

### 6.2 - Box Plot (Diagrama de Cajas)

Los **box plots** muestran la distribución de datos mediante cuartiles:
- **Mediana**: Línea central de la caja
- **Q1 y Q3**: Primer y tercer cuartil (bordes de la caja)
- **IQR**: Rango intercuartílico = Q3 - Q1
- **Whiskers**: Extienden hasta 1.5 × IQR
- **Outliers**: Puntos fuera de los whiskers

````python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

x = np.linspace(0, 10, 100)

# Subplot 1
axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title('sin(x)')

# Subplot 2
axes[0, 1].plot(x, np.cos(x), color='red')
axes[0, 1].set_title('cos(x)')

# Subplot 3
axes[1, 0].plot(x, x**2, color='green')
axes[1, 0].set_title('x²')

# Subplot 4
axes[1, 1].plot(x, np.log(x + 1), color='orange')
axes[1, 1].set_title('log(x+1)')

plt.tight_layout()
plt.show()
````
