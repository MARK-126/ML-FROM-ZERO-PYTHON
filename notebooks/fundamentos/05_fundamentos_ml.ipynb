## 2 - ¿Qué es Machine Learning?

**Machine Learning** es la ciencia de programar computadoras para que aprendan de datos sin ser explícitamente programadas para cada tarea específica.

### 2.1 - Paradigma Tradicional vs Machine Learning

**Programación Tradicional:**
$$\text{Datos} + \text{Reglas (código)} \rightarrow \text{Respuestas}$$

Ejemplo: Programa que suma dos números
```python
def suma(a, b):
    return a + b  # Regla explícita
```

**Machine Learning:**
$$\text{Datos} + \text{Respuestas} \rightarrow \text{Reglas (Modelo)}$$

Ejemplo: Modelo que aprende a predecir precios de casas
```python
modelo.fit(X_caracteristicas_casas, y_precios)  # Aprende las reglas
precio_predicho = modelo.predict(nueva_casa)
```

### 2.2 - Tipos de Machine Learning

#### 🎯 1. Aprendizaje Supervisado

**Definición:** Tenemos datos etiquetados $(X, y)$ donde $X$ son las features y $y$ es el target.

**Objetivo:** Aprender una función $f$ tal que:
$$y = f(X) + \epsilon$$

donde $\epsilon$ es el error irreducible.

**Tipos:**
- **Regresión**: Predecir valores continuos
  - Ejemplos: precio de casa, temperatura, ventas
  - Algoritmos: Regresión Lineal, Random Forest Regressor
  
- **Clasificación**: Predecir categorías discretas
  - Ejemplos: spam/no spam, gato/perro, enfermo/sano
  - Algoritmos: Regresión Logística, SVM, Random Forest Classifier

#### 🔍 2. Aprendizaje No Supervisado

**Definición:** Solo tenemos datos $X$ (sin etiquetas $y$).

**Objetivo:** Encontrar patrones, estructura o relaciones ocultas en los datos.

**Tipos:**
- **Clustering**: Agrupar datos similares
  - Ejemplos: Segmentación de clientes, detección de comunidades
  - Algoritmos: K-Means, DBSCAN, Hierarchical Clustering
  
- **Reducción de dimensionalidad**: Comprimir datos preservando información
  - Ejemplos: Visualización, compresión
  - Algoritmos: PCA, t-SNE, Autoencoders

#### 🎮 3. Aprendizaje por Refuerzo

**Definición:** Un agente aprende interactuando con un ambiente.

**Componentes:**
- **Estado**: Situación actual del agente
- **Acción**: Decisión que toma el agente
- **Recompensa**: Feedback del ambiente
- **Política**: Estrategia del agente ($\pi: \text{estado} \rightarrow \text{acción}$)

**Ejemplos:** AlphaGo, robótica, vehículos autónomos

## 3 - Terminología Fundamental

Estos términos son esenciales para comunicarte en el mundo de ML:

| Término | Descripción | Ejemplo |
|---------|-------------|---------|
| **Features** (X) | Variables de entrada / predictores | [tamaño_casa, num_habitaciones, ubicación] |
| **Target** (y) | Variable objetivo / a predecir | precio_casa |
| **Sample / Instance** | Una fila de datos | Una casa específica |
| **Dataset** | Conjunto completo de datos | Todas las casas |
| **Modelo** | Representación matemática aprendida | $\hat{y} = w_0 + w_1x_1 + w_2x_2$ |
| **Entrenamiento** | Proceso de aprender de los datos | `model.fit(X_train, y_train)` |
| **Predicción** | Usar el modelo en datos nuevos | `model.predict(X_new)` |
| **Parámetros** | Valores internos (aprendidos) | Pesos $w$ en regresión lineal |
| **Hiperparámetros** | Configuración (definidos por nosotros) | learning_rate, n_trees |

**Notación matemática común:**
- $n$ = número de muestras (samples)
- $m$ o $d$ = número de features
- $X \in \mathbb{R}^{n \times m}$ = matriz de features
- $y \in \mathbb{R}^{n}$ = vector de targets
- $\hat{y}$ = predicción del modelo

## 4 - Train/Test Split

Para evaluar si nuestro modelo **generaliza** a datos nuevos, dividimos los datos en conjuntos separados:

**División típica:**
$$\text{Dataset} \rightarrow \begin{cases} 
\text{Training Set (70-80\%)} & \text{Para entrenar} \\
\text{Test Set (20-30\%)} & \text{Para evaluar}
\end{cases}$$

**¿Por qué es necesario?**

1. **Evitar data leakage**: El modelo NO debe ver los datos de test durante entrenamiento
2. **Evaluar generalización**: Medir rendimiento en datos que el modelo nunca vió
3. **Simular producción**: El test set simula datos futuros/nuevos

**Regla de oro:** 🚨 **NUNCA** uses el test set para tomar decisiones sobre el modelo. Es solo para evaluación final.

### 4.1 - Implementación de Train/Test Split

# Implementación de train_test_split desde cero
def train_test_split_manual(X, y, test_size=0.2, random_state=None):
    """
    Divide datos en train y test sets.
    
    Parameters
    ----------
    X : ndarray
        Features de forma (n_samples, n_features) o (n_samples,)
    y : ndarray
        Target de forma (n_samples,)
    test_size : float
        Proporción del test set (0.0 a 1.0)
    random_state : int, optional
        Semilla aleatoria para reproducibilidad
    
    Returns
    -------
    tuple
        (X_train, X_test, y_train, y_test)
    """
    if random_state is not None:
        np.random.seed(random_state)
    
    n = len(X)
    
    # Crear índices aleatorios
    indices = np.random.permutation(n)
    
    # Calcular tamaño del test
    test_size_n = int(n * test_size)
    
    # Dividir índices
    test_indices = indices[:test_size_n]
    train_indices = indices[test_size_n:]
    
    # Dividir datos
    X_train = X[train_indices]
    X_test = X[test_indices]
    y_train = y[train_indices]
    y_test = y[test_indices]
    
    return X_train, X_test, y_train, y_test

# Ejemplo de uso
X = np.arange(100).reshape(-1, 1)  # 100 samples, 1 feature
y = np.arange(100) * 2 + 5          # y = 2x + 5

X_train, X_test, y_train, y_test = train_test_split_manual(
    X, y, test_size=0.2, random_state=42
)

print(f"📊 Tamaño total: {len(X)} samples")
print(f"📊 Training set: {len(X_train)} samples ({len(X_train)/len(X)*100:.0f}%)")
print(f"📊 Test set: {len(X_test)} samples ({len(X_test)/len(X)*100:.0f}%)")
print(f"\n✓ Primeros 5 índices de train: {X_train[:5].flatten()}")
print(f"✓ Primeros 5 índices de test: {X_test[:5].flatten()}")

## 5 - Overfitting y Underfitting

Estos son los dos problemas más comunes en Machine Learning:

### 5.1 - Underfitting (Alto Sesgo / High Bias)

**Síntomas:**
- El modelo es demasiado simple
- NO captura los patrones en los datos
- **Mal** rendimiento en training set
- **Mal** rendimiento en test set

**Soluciones:**
- Usar un modelo más complejo
- Agregar más features
- Reducir regularización

### 5.2 - Overfitting (Alta Varianza / High Variance)

**Síntomas:**
- El modelo es demasiado complejo
- Memoriza el training set (incluido el ruido)
- **Excelente** rendimiento en training set
- **Mal** rendimiento en test set

**Soluciones:**
- Usar un modelo más simple
- Obtener más datos de entrenamiento
- Aplicar regularización (L1, L2, Dropout)
- Usar validación cruzada

### 5.3 - Balance Ideal

**Objetivo:**
- Buen rendimiento en training set
- Buen rendimiento en test set
- El modelo **generaliza** correctamente

**Bias-Variance Tradeoff:**
$$\text{Error Total} = \text{Bias}^2 + \text{Variance} + \text{Irreducible Error}$$

## 6 - Métricas de Regresión

Las métricas nos permiten cuantificar qué tan bueno es nuestro modelo.

### 6.1 - Mean Squared Error (MSE)

**Fórmula:**
$$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

**Características:**
- Penaliza fuertemente errores grandes (por el cuadrado)
- Siempre no negativo
- Misma unidad que $y^2$
- Sensible a outliers

### 6.2 - Root Mean Squared Error (RMSE)

**Fórmula:**
$$\text{RMSE} = \sqrt{\text{MSE}} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$$

**Características:**
- Misma unidad que $y$ (más interpretable)
- Penaliza errores grandes

### 6.3 - Mean Absolute Error (MAE)

**Fórmula:**
$$\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$

**Características:**
- Penaliza todos los errores por igual
- Más robusta a outliers que MSE
- Misma unidad que $y$

### 6.4 - R² Score (Coeficiente de Determinación)

**Fórmula:**
$$R^2 = 1 - \frac{\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}{\sum_{i=1}^{n}(y_i - \bar{y})^2} = 1 - \frac{\text{SS}_{\text{res}}}{\text{SS}_{\text{tot}}}$$

**Interpretación:**
- $R^2 = 1$: Predicción perfecta
- $R^2 = 0$: Modelo tan bueno como predecir la media
- $R^2 < 0$: Modelo peor que predecir la media
- $R^2 \in [0, 1]$ típicamente: Proporción de varianza explicada

# Implementación de métricas de regresión
def mean_squared_error(y_true, y_pred):
    """
    Calcula el Error Cuadrático Medio (MSE).
    
    Parameters
    ----------
    y_true : ndarray
        Valores reales
    y_pred : ndarray
        Valores predichos
    
    Returns
    -------
    float
        MSE
    """
    return np.mean((y_true - y_pred) ** 2)

def mean_absolute_error(y_true, y_pred):
    """Calcula el Error Absoluto Medio (MAE)"""
    return np.mean(np.abs(y_true - y_pred))

def r2_score(y_true, y_pred):
    """
    Calcula el R² Score.
    
    R² = 1 indica predicción perfecta
    R² = 0 indica que el modelo es tan bueno como la media
    """
    ss_res = np.sum((y_true - y_pred) ** 2)  # Suma de cuadrados residual
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)  # Suma de cuadrados total
    return 1 - (ss_res / ss_tot)

# Ejemplo de uso
y_true = np.array([3, -0.5, 2, 7])
y_pred = np.array([2.5, 0.0, 2, 8])

print("📊 Valores reales:  ", y_true)
print("📊 Valores predichos:", y_pred)
print(f"\n✓ MSE:  {mean_squared_error(y_true, y_pred):.4f}")
print(f"✓ MAE:  {mean_absolute_error(y_true, y_pred):.4f}")
print(f"✓ R²:   {r2_score(y_true, y_pred):.4f}")

## 7 - Métricas de Clasificación

Para problemas de clasificación, usamos métricas diferentes:

### 7.1 - Accuracy (Exactitud)

**Fórmula:**
$$\text{Accuracy} = \frac{\text{Predicciones Correctas}}{\text{Total de Predicciones}} = \frac{TP + TN}{TP + TN + FP + FN}$$

**Cuándo usar:** Cuando las clases están balanceadas

**Limitación:** No es buena para datasets desbalanceados

### 7.2 - Matriz de Confusión

Para clasificación binaria (0/1, Negativo/Positivo):

```
                    Predicho
                 Neg (0)  Pos (1)
Real  Neg (0)      TN       FP
      Pos (1)      FN       TP
```

- **TN** (True Negative): Predijo negativo, era negativo ✓
- **TP** (True Positive): Predijo positivo, era positivo ✓
- **FP** (False Positive): Predijo positivo, era negativo ✗ (Error Tipo I)
- **FN** (False Negative): Predijo negativo, era positivo ✗ (Error Tipo II)

### 7.3 - Precision y Recall

**Precision (Precisión):**
$$\text{Precision} = \frac{TP}{TP + FP}$$

*"De todos los que predije como positivos, ¿cuántos realmente lo eran?"*

**Recall (Sensibilidad / Sensitivity):**
$$\text{Recall} = \frac{TP}{TP + FN}$$

*"De todos los positivos reales, ¿cuántos detecté?"*

**F1-Score (Media Armónica):**
$$F1 = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$

**Cuándo usar cada una:**
- **Precision alta**: Cuando FP son muy costosos (ej: spam detector)
- **Recall alto**: Cuando FN son muy costosos (ej: detector de cáncer)
- **F1**: Balance entre precision y recall

````python
def accuracy(y_true, y_pred):
    """Calcula la exactitud"""
    return np.mean(y_true == y_pred)

def confusion_matrix(y_true, y_pred):
    """
    Calcula la matriz de confusión para clasificación binaria.
    Returns: [[TN, FP], [FN, TP]]
    """
    TP = np.sum((y_true == 1) & (y_pred == 1))
    TN = np.sum((y_true == 0) & (y_pred == 0))
    FP = np.sum((y_true == 0) & (y_pred == 1))
    FN = np.sum((y_true == 1) & (y_pred == 0))
    return np.array([[TN, FP], [FN, TP]])

# Ejemplo
y_true = np.array([0, 1, 1, 0, 1, 1, 0, 0, 1, 0])
y_pred = np.array([0, 1, 1, 0, 0, 1, 0, 1, 1, 0])

print(f"Accuracy: {accuracy(y_true, y_pred):.2f}")
print(f"\nMatriz de Confusión:")
print(confusion_matrix(y_true, y_pred))
````
