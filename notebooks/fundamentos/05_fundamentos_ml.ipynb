## 1. ¿Qué es Machine Learning?

**Machine Learning** es la ciencia de programar computadoras para que aprendan de datos sin ser explícitamente programadas.

### Paradigma Tradicional vs ML

**Programación Tradicional:**
- Datos + Reglas → Respuestas

**Machine Learning:**
- Datos + Respuestas → Reglas (Modelo)

### Tipos de Machine Learning

#### 1. **Aprendizaje Supervisado**
- Tenemos datos etiquetados (X, y)
- El modelo aprende la relación entre X e y
- Ejemplos: 
  - **Regresión**: Predecir valores continuos (precio de casa, temperatura)
  - **Clasificación**: Predecir categorías (spam/no spam, gato/perro)

#### 2. **Aprendizaje No Supervisado**
- Solo tenemos datos X (sin etiquetas)
- El modelo encuentra patrones
- Ejemplos:
  - **Clustering**: Agrupar datos similares
  - **Reducción de dimensionalidad**: PCA

#### 3. **Aprendizaje por Refuerzo**
- El agente aprende interactuando con el ambiente
- Recibe recompensas/penalizaciones
- Ejemplos: Juegos, robótica, autos autónomos

## 2. Terminología Fundamental

- **Features (Características)**: Variables de entrada (X)
- **Target (Objetivo)**: Variable a predecir (y)
- **Muestra/Sample**: Una fila de datos
- **Modelo**: La representación matemática aprendida
- **Entrenamiento**: Proceso de aprender de los datos
- **Predicción**: Usar el modelo en datos nuevos
- **Parámetros**: Valores internos del modelo (aprendidos)
- **Hiperparámetros**: Configuración del modelo (definidos por nosotros)

## 3. Train/Test Split

Para evaluar nuestro modelo, dividimos los datos en:
- **Training Set**: Para entrenar el modelo (típicamente 70-80%)
- **Test Set**: Para evaluar el modelo (típicamente 20-30%)

**¿Por qué?** Queremos saber si el modelo generaliza a datos nuevos, no solo memoriza el entrenamiento.

````python
def train_test_split_manual(X, y, test_size=0.2, random_state=None):
    """
    Divide datos en train y test.
    
    Parámetros:
    -----------
    X : array
        Features
    y : array
        Target
    test_size : float
        Proporción del test (0.0 a 1.0)
    random_state : int
        Semilla aleatoria
    
    Returns:
    --------
    X_train, X_test, y_train, y_test
    """
    if random_state is not None:
        np.random.seed(random_state)
    
    # Número de muestras
    n = len(X)
    
    # Índices aleatorios
    indices = np.random.permutation(n)
    
    # Calcular tamaño del test
    test_size_n = int(n * test_size)
    
    # Dividir índices
    test_indices = indices[:test_size_n]
    train_indices = indices[test_size_n:]
    
    # Dividir datos
    X_train = X[train_indices]
    X_test = X[test_indices]
    y_train = y[train_indices]
    y_test = y[test_indices]
    
    return X_train, X_test, y_train, y_test

# Ejemplo
X = np.arange(100).reshape(-1, 1)
y = np.arange(100)

X_train, X_test, y_train, y_test = train_test_split_manual(X, y, test_size=0.2, random_state=42)

print(f"Tamaño total: {len(X)}")
print(f"Tamaño train: {len(X_train)}")
print(f"Tamaño test: {len(X_test)}")
````

## 4. Overfitting y Underfitting

### Underfitting (Sesgo Alto)
- El modelo es demasiado simple
- No captura los patrones de los datos
- Mal rendimiento en train Y test

### Overfitting (Varianza Alta)
- El modelo es demasiado complejo
- Memoriza el training set
- Excelente en train, malo en test

### Balance Ideal
- Buen rendimiento en ambos conjuntos
- El modelo generaliza bien

````python
# Visualización de Overfitting
np.random.seed(42)
X = np.linspace(0, 10, 50)
y_true = 2 * X + 1
y = y_true + np.random.randn(50) * 2

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Underfitting - modelo muy simple (constante)
axes[0].scatter(X, y, alpha=0.5)
axes[0].axhline(y=np.mean(y), color='red', linewidth=2, label='Predicción')
axes[0].set_title('UNDERFITTING\nModelo muy simple')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Balance correcto - modelo lineal
coeffs = np.polyfit(X, y, 1)
y_pred = np.polyval(coeffs, X)
axes[1].scatter(X, y, alpha=0.5)
axes[1].plot(X, y_pred, color='green', linewidth=2, label='Predicción')
axes[1].set_title('BALANCE CORRECTO\nModelo apropiado')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Overfitting - polinomio de grado alto
coeffs = np.polyfit(X, y, 15)
y_pred = np.polyval(coeffs, X)
axes[2].scatter(X, y, alpha=0.5)
axes[2].plot(X, y_pred, color='orange', linewidth=2, label='Predicción')
axes[2].set_title('OVERFITTING\nModelo muy complejo')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## 5. Métricas de Evaluación

### Para Regresión:

**Mean Squared Error (MSE):**
$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

**Mean Absolute Error (MAE):**
$$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$

**R² Score (Coeficiente de Determinación):**
$$R^2 = 1 - \frac{\sum(y_i - \hat{y}_i)^2}{\sum(y_i - \bar{y})^2}$$

R² = 1 es perfecto, R² = 0 es igual a predecir la media.

````python
def mean_squared_error(y_true, y_pred):
    """Calcula el Error Cuadrático Medio"""
    return np.mean((y_true - y_pred) ** 2)

def mean_absolute_error(y_true, y_pred):
    """Calcula el Error Absoluto Medio"""
    return np.mean(np.abs(y_true - y_pred))

def r2_score(y_true, y_pred):
    """Calcula el R² Score"""
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    return 1 - (ss_res / ss_tot)

# Ejemplo
y_true = np.array([3, -0.5, 2, 7])
y_pred = np.array([2.5, 0.0, 2, 8])

print(f"MSE: {mean_squared_error(y_true, y_pred):.4f}")
print(f"MAE: {mean_absolute_error(y_true, y_pred):.4f}")
print(f"R²: {r2_score(y_true, y_pred):.4f}")
````

### Para Clasificación:

**Accuracy (Exactitud):**
$$Accuracy = \frac{\text{Predicciones Correctas}}{\text{Total de Predicciones}}$$

**Matriz de Confusión:**
```
                Predicho
              Neg    Pos
Real  Neg     TN     FP
      Pos     FN     TP
```

**Precision:** De todos los que predije como positivos, ¿cuántos realmente lo eran?
$$Precision = \frac{TP}{TP + FP}$$

**Recall:** De todos los positivos reales, ¿cuántos detecté?
$$Recall = \frac{TP}{TP + FN}$$

**F1-Score:** Media armónica de Precision y Recall
$$F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}$$

````python
def accuracy(y_true, y_pred):
    """Calcula la exactitud"""
    return np.mean(y_true == y_pred)

def confusion_matrix(y_true, y_pred):
    """
    Calcula la matriz de confusión para clasificación binaria.
    Returns: [[TN, FP], [FN, TP]]
    """
    TP = np.sum((y_true == 1) & (y_pred == 1))
    TN = np.sum((y_true == 0) & (y_pred == 0))
    FP = np.sum((y_true == 0) & (y_pred == 1))
    FN = np.sum((y_true == 1) & (y_pred == 0))
    return np.array([[TN, FP], [FN, TP]])

# Ejemplo
y_true = np.array([0, 1, 1, 0, 1, 1, 0, 0, 1, 0])
y_pred = np.array([0, 1, 1, 0, 0, 1, 0, 1, 1, 0])

print(f"Accuracy: {accuracy(y_true, y_pred):.2f}")
print(f"\nMatriz de Confusión:")
print(confusion_matrix(y_true, y_pred))
````
